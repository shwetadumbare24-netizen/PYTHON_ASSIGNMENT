This is a single Inheritance:

```python
class Parent:
    # def __init__(self):
    #     pass
    def Parent_method(self):
        print("Parent Classs")
class Child(Parent):
    def Child_method(self):
        print("Child Classs")

# obj1 = Parent()
# obj1.Parent_method()

obj2 = Child()
obj2.Child_method()
obj2.Parent_method()
```

Constructor With Single Inheritance

```python
class Parent:
    def __init__(self):
        print("this is parent")

    def Parent_method(self):
        print("Parent")

class Child(Parent):

    def __init__(self):
            print("this is child class constructor")
    def Child_method(self):
        print("Child Classs")

obj1 = Child()
# obj1.Child_method()
```

```python
class one:
    def greet(self):
        print("hello from parent")

class two(one):
    def greet(self):
        super().__init__()
        print("hello from child class")
obj1 = two()
obj1.greet()
```
