Constructor With Single Inheritance

```python
class Parent:
    def __init__(self):
        print("this is parent")

    def Parent_method(self):
        print("Parent")

class Child(Parent):

    def __init__(self):
            print("this is child class constructor")
    def Child_method(self):
        print("Child Classs")

obj1 = Child()
# obj1.Child_method()
```

```python
class one:
    def greet(self):
        print("hello from parent")

class two(one):
    def greet(self):
        super().__init__()
        print("hello from child class")
obj1 = two()
obj1.greet()
```

```python
class Bank:
    def __init__(self):
        self.name = "SONU"
        self.acc_num = 876596875858658
        self.__pin = 12345

    # getter method used for access private data
    def get_pin(self):
        return self.__pin

    #setter method used for updated private data
    def set_pin(self,pin):
        self.__pin = pin
    
class subBank(Bank):
    def display(self):
        print("User Name:", self.name)
        print("Account Number:", self.acc_num)
        print("User pin:", self.get_pin())
        print( "updated pin:", self.set_pin())

obj1 = subBank()
obj1.display()

print(obj1.get_pin())
print(obj1.set_pin(1234567789))
```
