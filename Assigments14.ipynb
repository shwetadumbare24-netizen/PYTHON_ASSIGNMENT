1. Student Information System Create a Student class with: • name  • roll_no  • branch  • marks  Create 3 student objects and create a method display() to print their details

```python
class student:
    def __init__(self,name,roll_no,branch,marks):
        self.name = name
        self.roll_no = roll_no
        self.branch = branch
        self.marks = marks
    def display_info(self):
        print(self.name,self.roll_no,self.branch,self.marks)

obj1 = student("Sonu", 21, "AIML", 90)
obj2 = student("Shubhazz", 22, "CO" , 95)
obj3 = student("Balasvi", 20 , "AIML" , 99)
obj1.display_info()
obj2.display_info()
obj3.display_info()


```

2. Employee Salary Create an Employee class with:      
• name  • employee_id  • salary  
Create a method display_salary() that displays the employee's salary. 
Create 3 employee objects. : Don't assign the values directly inside the class. 
Pass them while creating objects. 

```python
class employee:
    def __init__(self,Emp_name,Emp_id,salary):
        self.Emp_name = Emp_name
        self.Emp_id = Emp_id
        self.salary = salary
    
    def display_info(self):
        print(self.Emp_name,self.Emp_id,self.salary)

obj1 = employee("Sarthak", 101, 90000)
obj2 = employee("Swapnil", 102, 950000)
obj3 = employee("Prathmesh", 103, 99000)
obj1.display_info()
obj2.display_info()
obj3.display_info()
```

3. Mobile Phone Create a Mobile class with: 

• brand  • model  • price  

Create a method: display_details()

Create 3 mobile objects with different information.

 Expected concept: Constructor + objects. 

```python
class mobile:
    def __init__(self,brand,model,price):
        self.brand = brand
        self.model = model
        self.price = price
    def display(self):
        print(self.brand,self.model,self.price)
        

obj1 = mobile("Samsung", "Galaxy S21", 799)
obj2 = mobile("Apple", "iPhone 13", 999)
obj3 = mobile("OnePlus", "9 Pro", 969)
obj1.display()
obj2.display()
obj3.display()
```
