1. Write a program to count the frequency of each character in a string using a dictionary. - - -  - - Input: "programming" Output: {'p':1, 'r':2, 'o':1, ...} 


```python
s = "Programming"
d = {}

for i in s:
    d[i]=s.count(i)
print(d)
```

2. Write a program to count the frequency of each element in a list using a dictionary.  [10,20,10,30,20,10] 
Output: {10:3, 20:2, 30:1} 


```python
number = [10,20,10,30,20,10] 
d = {}
for i in number:
    d[i]=number.count(i)
print(d)
```

3. Write a program to find the key having the maximum value in a dictionary.
{'A':85, 'B':92, 'C':78}

```python
number = {'A':85, 'B':92, 'C':78} 
key = None
maxvalue = 0
for i in number:
  
    if number[i] > maxvalue:
        maxvalue = number[i]
        key = i

print(key)
```

6. Write a program to remove duplicate values from a dictionary. - - -  - - {'a':10, 'b':20, 'c':10, 'd':30} Output: {'a':10, 'b':20, 'd':30} 

```python
number = {'a':10, 'b':20, 'c':10, 'd':30}
new = {}
for key,value in number.items():
    if value not in new.values():
        new[key]=value
print(new) 
```

8. Write a program to find the sum of all values in a dictionary. - {'Math':85, 'Science':90, 'English':80} -  - - Output: 255 

```python
number = {'Math':85, 'Science':90, 'English':80}

total = 0

for i in number:
    total = total + number[i]
print(total)
```

9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. - - Keys = ['A', 'B', 'C'] Values = [10, 20, 30] 

```python
keys = ['A', 'B', 'C']
values = [10, 20, 30]
d = {}

for i in range(len(keys)):
    d[keys[i]] = values[i]

print(d)
```

10. Write a program to create a student result dictionary containing name and marks of 5 students, then display the topper.

```python
students = {'A': 85, 'B': 92, 'C': 78, 'D': 88, 'E': 95 }

topper = None
maxmarks = 0

for i in students:
    if students[i] > maxmarks:
        maxmarks = students[i]
        topper = i
print("Topper:",topper)
print("Maxmarks:",maxmarks)

 
```
