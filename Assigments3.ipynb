2. Take the length and width of a rectangle as input and calculate: - - Perimeter 

```python
length = float(input("Enter Length: "))
width = float(input("Enter Width: "))

area = length * width
perimeter = 2 * (length + width)

print("Area =", area)
print("Perimeter =", perimeter)
```

3. Take the price and quantity of a product from the user and calculate: - Total Amount - - GST (18%) Final Bill 

```python
price = float(input("Enter Product Price: "))
quantity = int(input("Enter Quantity: "))

total = price * quantity
gst = total * 0.18
final_bill = total + gst

print("Total Amount =", total)
print("GST =", gst)
print("Final Bill =", final_bill)
```

7. Write a program to determine whether a year is a Leap Year. 

```python
year = int(input("Enter Year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

8. Create a simple calculator using match-case.
 - Operations:
               • Addition  
               • Subtraction
               • Multiplication 
               • Division 

```python
a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))

choice = int(input("""
1. Addition
2. Subtraction
3. Multiplication
4. Division
Enter Choice: """))

match choice:
    case 1:
        print("Addition =", a + b)
    case 2:
        print("Subtraction =", a - b)
    case 3:
        print("Multiplication =", a * b)
    case 4:
        print("Division =", a / b)
    case _:
        print("Invalid Choice")
```
