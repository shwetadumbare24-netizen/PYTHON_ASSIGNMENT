```python
class car:
    def __init__(self):
        self.name = "Car"
        self.color = "Red"
        self.model = "Sedan"
        self.milage = "Milage"
    def display(self):
        print("Name:", self.name)
        print("Color:", self.color)
        print("Model:", self.model)
        print("Milage:", self.milage)
car1 = car()
car1.display()
```

```python
class student:
    def __init__(self,name,age):
        self.name= name
        self.age = age
    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
s1 = student("Alice", 20)
s2 = student("Bob",22)
s1.display()
s2.display()
```
