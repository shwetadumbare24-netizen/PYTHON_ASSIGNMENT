```python
class car:
    def __init__(self):
        self.name = "Car"
        self.color = "Red"
        self.model = "Sedan"
        self.milage = "Milage"
    def display(self):
        print("Name:", self.name)
        print("Color:", self.color)
        print("Model:", self.model)
        print("Milage:", self.milage)
car1 = car()
car1.display()
```

```python
def MoveZeroes(*nums):
    arr1 = []
    arr2 = []

    for i in nums:
        if i == 0:
            arr1.append(i)
        else:
            
            arr2.append(i)
          
    res = arr2+arr1
    return res

print(MoveZeroes(0,1,0,3,12))



```
