```python
class demo:
    def __init__(self,num1):
        self.numone = num1

    def __str__(self):
        return self.numone

    def __len__(self):
        return len(self.numone)

    def __eq__(value1,value2):
        return value1.numone == value2.numone

obj1 = demo("10")
obj2 = demo("20")
print(obj1)
print(len(obj1))
print(obj1==obj2)
```

```python
class Temp:
    def __init__(self,num1):
        self.numone = num1

    def __add__(value1,value2):
        return value1.numone + value2.numone
    def __sub__(value1,value2):
        return value1.numone - value2.numone
    def __mul__(num1,num2):
        return num1.numone * num2.numone 
    def __truediv__(num1,num2):
        return num1.numone / num2.numone
    def __floordiv__(num1,num2):
        return num1.numone // num2.numone

    def __mod__(num1,num2):
        return num1.numone % num2.numone

     

    

       
obj1 = Temp(60)
obj2 = Temp(20)
print(obj1 + obj2)
print(obj1 - obj2)
print(obj1 * obj2)
print(obj1 / obj2)
print(obj1 // obj2)
print(obj1 % obj2)
```
