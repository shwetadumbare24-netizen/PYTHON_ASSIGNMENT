```python
#Count the Number of Vowels and Consonants

string = input("Enter a string:")

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0

for sh in string:
    if sh.isalpha():
        if sh in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
print("Vowels:" , vowel_count)
print("Consonant:", consonant_count)
```

```python
#Check Whether a String is Palindrome

string = input("Enter a string:")
reverse = ""

for ch in string:
    reverse = ch + reverse

if string == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Reverse a String Without Slicing or Built-in Functions
string = input("enter a string:")

reverse = ""
for ch in string:
    reverse = ch + reverse

print("Reversed string:",reverse)

```

```python
#Count the Occurrence of Each Character
string = input("Enter a string:")

visited = ""

for ch in string:
    if ch not in visited:
        count = 0
        for i in string:
            if ch == i:
                count += 1
        print(ch,":",count)
        visited += ch
```

```python
#Remove Duplicate Characters While Maintaining Order
string = input("Enter a string:")

result = ""

for ch in string:
    if ch not in result:
        result += ch

print("output:", result)
```
