1. Write a program to count the number of vowels and consonants in a given string. 

```python
string = input("Enter a string:")

vowels = "aeiouAEIOU"
vowel_count = 0
consonant_count = 0

for sh in string:
    if sh.isalpha():
        if sh in vowels:
            vowel_count += 1
        else:
            consonant_count += 1
print("Vowels:" , vowel_count)
print("Consonant:", consonant_count)
```

2. Write a program to check whether a given string is a Palindrome. - - Input: madam Output : Palindrome 

```python
string = input("Enter a string:")
reverse = ""

for ch in string:
    reverse = ch + reverse

if string == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
string = input("enter a string:")

reverse = ""
for ch in string:
    reverse = ch + reverse

print("Reversed string:",reverse)

```

4. Write a program to count the occurrence of each character in a string. Input : programming Output: p : 1 r : 2 o : 1 ... 

```python
string = input("Enter a string:")

visited = ""

for ch in string:
    if ch not in visited:
        count = 0
        for i in string:
            if ch == i:
                count += 1
        print(ch,":",count)
        visited += ch
```

5. Write a program to remove all duplicate characters from a string while maintaining their original order. - - Input : programming Output : progamin 

```python
string = input("Enter a string:")

result = ""

for ch in string:
    if ch not in result:
        result += ch

print("output:", result)
```

6. Write a program to find the longest word in a sentence. - Input: Python is an amazing programming language  - Output: programming 

```python
sentence = input("Enter a sentence: ")

words = sentence.split()      

longest = ""                  

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word is:", longest)
```

7. Write a program to capitalize the first letter of every word without using the title() method. - - Input:    welcome to python Output:   Welcome To Python 

```python
text = input("Enter a sentence: ")

words = text.split()

for word in words:
    print(word.capitalize(), end=" ")
```

9. Write a program to remove all spaces from a string. - Python Programming Language 

```python
text = input("Enter a string: ")

result = ""

for ch in text:
    if ch != " ":
        result += ch

print("Output:", result)
```
