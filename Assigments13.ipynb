6. Write a program to read a file and create another file containing only the lines that contain the word "Python".

```python
with open("input.txt", "r") as file:
    with open("python_lines.txt", "w") as new_file:
        for line in file:
            if "python" in line:
                new_file.write(line)
print("File is created successfully.")
                

```

8. Write a program to read a file and find the longest word present in it.  

```python
with open("input.txt","r") as file:
    words = file.read().split()

longest_word = max(words, key=len)
print("longest word is:",longest_word)
```

9. Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.  

```python
file = open("student1.txt", "w")

file.write("10\n")
file.write("20\n")
file.write("30\n")
file.write("40\n")
file.write("50\n")

file.close()
```

```python
file = open("student1.txt", "r")
numbers = file.read().split()

numbers = [int(num) for num in numbers]

total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print("sum:", total)
print("average:", average)
print("maximum:", maximum)
print("minimum:", minimum)

file.close()
```
