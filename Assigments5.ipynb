1. Write a program to count the number of even and odd numbers in a list. 

```python
numbers = [10, 15, 20, 25, 30, 35]

even = 0
odd = 0

for i in numbers:
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

2. Write a program to create a new list containing only the positive numbers from a given list. - - -  - - Input: [-10, 20, -5, 40, 0, 15] Output: [20, 40, 15] 

```python
numbers = [-10, 20, -5, 40, 0, 15]

positive = []

for i in numbers:
    if i > 0:
        positive.append(i)

print(positive)
```

3. Write a program to remove all duplicate elements from a list while preserving the original order. - - -  - - Input: [10,20,30,20,40,10,50] Output: [10,20,30,40,50] 

```python
numbers = [10, 20, 30, 20, 40, 10, 50]

new = []

for i in numbers:
    if i not in new:
        new.append(i)

print(new)
```

4. Write a program to find the second largest element in a list. - - -  - - [12,45,67,89,54] Output: 67 

```python
numbers = [12, 45, 67, 89, 54]

numbers.sort()

print("Second Largest Element:", numbers[-2])
```

6. Write a program to count the frequency of every element in a list. - Input: - -  - - - - [1,2,3,2,1,4,1] Output: 1 -> 3 2 -> 2 3 -> 1 

```python
numbers = [1, 2, 3, 2, 1, 4, 1]

for i in numbers:
    print(i, "->", numbers.count(i))
```

7. Write a program to reverse a list without using the reverse() method or slicing ([::-1]). 

```python
numbers = [10, 20, 30, 40, 50]

reverse_list = []

for num in numbers:
    reverse_list.insert(0, num)

print("Reversed List:", reverse_list)
```

9. Write a program to find all common elements between two lists. 
- List1 = [10,20,30,40] 
- List2 = [20,40,60] 
Output: [20,40] 

```python
list1 = [10, 20, 30, 40]
list2 = [20, 40, 60]

common = []

for i in list1:
    if i in list2:
        common.append(i)

print("Common Elements:", common)
```

10. Write a program to separate all even and odd numbers into two different lists.
[10,15,20,25,30] 
Output: Even = [10,20,30] Odd = [15,25] 

```python
numbers = [10, 15, 20, 25, 30]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even =", even)
print("Odd =", odd)
```
