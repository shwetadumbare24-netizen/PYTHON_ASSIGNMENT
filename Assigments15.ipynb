2. Product Class – __eq__() 

Create a Product class with name and price.
 
Implement __eq__() to check whether two products have the same price. 

```python
class Product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def __eq__(self,other):
        return self.name == other.price
p1 = Product("MacBook", 150000)
p2 = Product("Lenevo",65000)
print(p1==p2)

```

3. Shopping Cart – __len__() 

Create a ShoppingCart class containing a list of products. 

Implement __len__() so that len(cart) returns the number of products in the cart. 

```python
class ShoppingCart:
    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)


cart = ShoppingCart(["Laptop", "Mobile", "Mouse", "Divya"])

print(len(cart))
```

4. Bank Account – __str__() 

Create a BankAccount class with account holder name, account number, and balance. 

Implement __str__() to display account information when the object is printed. 

```python
class BankAccount:
    def __init__(self,name,acc_no,balance):
        self.name = name
        self.acc_no = acc_no
        self.balance = balance

    def __str__(self):
        return f"Account Holder: {self.name} \nAccount Number: {self.acc_no} \nBalance: ₹{self.balance}"
b1 = BankAccount("shwetali","98765895478954",100000)
print(b1)
```

5. Number Class – __add__() Create a Number class that stores one number.

         Implement __add__() so that two objects can be added 

```python
class add:
    def __init__(self,num1):
        self.num1 = num1
    def __add__(self,other):
        return self.num1 + other.num1
add1 = add(20)
add2 = add(40)
print(add1 + add2)
```

6. Create a Number class and implement: 
1. __add__() 
2. __sub__() 
3. __mul__()  
4. __truediv__()  
5. __mod__() 

```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num

    def __sub__(self, other):
        return self.num - other.num

    def __mul__(self, other):
        return self.num * other.num

    def __mod__(self, other):
        return self.num % other.num


obj1 = Number(20)
obj2 = Number(6)

print(obj1 + obj2)   
print(obj1 - obj2)   
print(obj1 * obj2)   
print(obj1 % obj2)   
```
