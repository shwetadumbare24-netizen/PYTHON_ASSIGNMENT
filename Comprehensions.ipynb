```python
listOne = [1,2,3,4,5,6,7,8]
listTwo = []

for i in listOne:
    listTwo.append(i*2)
print("List One:",listOne)
print("List Two:",listTwo)
```

```python
listOne = [1,2,3,4,5,6,7,8]
listTwo = []

for i in listOne:
    if i % 2 == 0:
        listTwo.append(i**2)
print("List One:",listOne)
print("List Two:",listTwo)
```

```python
listOne = [1,2,3,4,5,6,7,8]

# [expression,iteration,condition(optional)]
# [    3    ,     1    ,     2            ]
listTwo = [i**2 for i in listOne if i%2 == 0]

print("Origianl list:",listOne)
print("Square List:",listTwo)
```

```python
listOne = [1,3,5,9,8,7,4]

listTwo = [i*2 for i in listOne if i%2 == 0]
print("List One :",listOne)
print("List Two :",listTwo)
```

Set Compression

```python
setOne = {1,2,3,4,5,6,7,8,9}

setTwo = {i**2 for i in setOne if i%2 == 0}
#---------expression       , iteration    , condition ------------------#
print("set One :",setOne)
print("set Two :",setTwo)

```

Dictionary Compression

```python
listOne = ["mohit","rohit","ram","sham"]

DictOne = {i:len(i) for i in listOne}#key :value
print(DictOne)
```

```python
listOne = [1,2,3,4,5,6,7,8]

dictOne = {i:i+i for i in listOne}
dict1 = {i:i**2 for i in listOne}
dict2 = {i:i**3 for i in listOne}
print("Dictionary For Addition:",dictOne)
print("Dictionary For Square:",dict1)
print("Dictinary For Cube:",dict2)
```
