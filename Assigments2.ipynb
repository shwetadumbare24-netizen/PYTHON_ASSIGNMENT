```python
# Primitive Data Types

integer_value = 25
float_value = 12.5
complex_value = 4 + 3j
string_value = "Python"
boolean_value = True

print("Integer")
print("Value:", integer_value)
print("Type:", type(integer_value))
print("ID:", id(integer_value))

print("\nFloat")
print("Value:", float_value)
print("Type:", type(float_value))
print("ID:", id(float_value))

print("\nComplex")
print("Value:", complex_value)
print("Type:", type(complex_value))
print("ID:", id(complex_value))

print("\nString")
print("Value:", string_value)
print("Type:", type(string_value))
print("ID:", id(string_value))

print("\nBoolean")
print("Value:", boolean_value)
print("Type:", type(boolean_value))
print("ID:", id(boolean_value))
```

```python
a = 15
b = 4

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Exponent =", a ** b)
```

```python
x = "100"
y = 50

x = int(x)

print("Addition =", x + y)
print("Subtraction =", x - y)
print("Multiplication =", x * y)
print("Division =", x / y)
```

```python
a = 10
b = 20

print(a + b * 2)
print((a + b) * 2)
print(a ** 2 + b)
print(a // 3)
print(a / 3)
```

```python
units = 250
cost_per_unit = 8

total_bill = units * cost_per_unit
gst = total_bill * 0.18
final_bill = total_bill + gst

print("Units Consumed:", units)
print("Cost Per Unit:", cost_per_unit)
print("Total Bill:", total_bill)
print("GST (18%):", gst)
print("Final Bill:", final_bill)
```

```python
languages = ["Python", "C", "C++", "JavaScript"]

print("Python" in languages)
print("Java" not in languages)
print("C++" in languages)
print("C+" in languages)
```

```python
# int → float
num = 50
float_num = float(num)
print(float_num)

# float → string
price = 99.99
string_price = str(price)
print(string_price)

# string → int
age = "20"
int_age = int(age)
print(int_age)

# int → bool
number = 1
bool_number = bool(number)
print(bool_number)

# bool → int
flag = True
int_flag = int(flag)
print(int_flag)
```

```python
product_name = "Laptop"
price = 50000
quantity = 2

total_cost = price * quantity
gst = total_cost * 0.18
final_amount = total_cost + gst

print("Product Name:", product_name)
print("Price:", price)
print("Quantity:", quantity)
print("Total Cost:", total_cost)
print("GST:", gst)
print("Final Amount:", final_amount)
```
