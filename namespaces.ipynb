NamesSpaces:

System used by python to store block of code where we can access a name.

```python
name = "Sonu"
age = 22

def temp():
    roll = 55    # local namespace variable
    mark = 75    # local namespace variable
    print("Global Variable:",name,age) #Access inside the function

  
    print("Local namespace Variable:", roll) #it can be aceess only inside the block of function
    print("Local namespace Variable:",mark)

temp()
print("Global namespace Variable:",name,age) #access outside the function
```

3.Enclosing Namesapce:-

enclosing namesapce is the namespace of an outer function that contains a nested inner function

```python
name = "Sonu" # global variable

def outer():
    age = 21  #  enclosing variable
    
    def inner():   # local variable
        marks = 95
        print(marks)
        print(age)
        print(name)
    inner()
outer()
```

Shadowing :-

- same variable in decalre two variable scopes,both will be printed
 

```python
name = "Sonu"

def temp():
    name = "Ravi"  # shadowing of global variable
    print(name)  # it will print local variable
temp()
print(name)
```

Global Keyword:


```python
name = "Sonu"

def temp():
    global name   #global Keyword is used to access the global variable inside the function
    name = "Ravi"  
    print(name)  
temp()
print(name)
```
