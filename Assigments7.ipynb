1. Write a program to find the union of two sets without using the union() method. 

```python
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}
print("Set1:",set1)
print("Set2:",set2)
result = set1

for i in set2:
    result.add(i)

print("Union =", result)
```

2. Write a program to remove all duplicate elements from a list using a set


```python
list1 = [10, 20, 30, 20, 40, 10, 50, 30]

res = list(set(list1))

print("Original List =", list1)
print("List after removing duplicates =", res)
```

3. Write a program to find all common elements between two lists using sets. 


```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
print("List1:",list1)
print("List2:",list2)
common = set(list1).intersection(set(list2))

print("Common Elements =", common)
```

4. Write a program to count the number of unique elements in a list using a set.

```python
list1 = [10, 20, 30, 20, 40, 10, 50, 30]

unique = set(list1)

print("Unique Elements =", unique)
print("Number of Unique Elements =", len(unique))
```

5. Write a program to find the missing numbers between 1 to N using sets.
 
 Input: N = 10 
 Numbers = {1,2,3,5,7,8,10} 
 Output: {4,6,9} 

```python
set1 = {1,2,3,5,7,8,10}
print("Set:",set1)

mising = set()

for i in range(1,11):
    if i not in set1:
        mising.add(i)
print("Missing Number:",mising)
```

6. Write a program to create a new set containing only the even numbers from a given set. 

```python
set1 = {1,2,3,5,7,8,10}
print("Set:",set1)

res = set()
for i in set1:
    if i % 2 == 0:
        res.add(i)

print("Set contaning only even element:",res)
```

7. Write a program to merge three sets into a single set containing only unique elements.

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
set3 = {6, 7, 8, 9}

result = set1.union(set2, set3)

print("Set 1:", set1)
print("Set 2:", set2)
print("Set 3:", set3)
print("Merged Set:", result)
```

9. Given multiple lists, find all elements that appear in every list using sets.

```python
list1 = [1, 2, 3, 4, 5]
list2 = [2, 3, 5, 6, 7]
list3 = [1, 2, 3, 5, 8]

set1 = set(list1)
set2 = set(list2)
set3 = set(list3)

common = set1.intersection(set2, set3)

print("Common Elements:", common)
```
