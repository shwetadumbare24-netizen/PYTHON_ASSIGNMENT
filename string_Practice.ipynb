```python
# Nested list means list inside a list
listOne = [1, 2, 3, [4, 5, 6], 7, 8, 9]

print(listOne)
print(listOne[3])

for i in listOne:
    if isinstance(i, list):   # Check if the element is a list
        for j in range(len(i)):
            print(i[j])
    else:
        print(i)
```
